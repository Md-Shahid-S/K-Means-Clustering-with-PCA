```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'../data/raw/Mall_Customers.csv')
```

```python
df.head()
```

```python
num_features = df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']].values
cat_features = df[['Gender']].values
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
num_scaled = scaler.fit_transform(num_features)
num_scaled[:5]
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder()
cat_encoded = encoder.fit_transform(cat_features).toarray()
cat_encoded[:5]
```

```python
X = np.concatenate([num_scaled, cat_encoded], axis=1)
print("Final shape:", X.shape)
X[:5]
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=0.95, random_state=45)
X_pca = pca.fit_transform(X)
print("PCA shape:", X_pca.shape)
X_pca[:5]
```

```python
X_pca_df = pd.DataFrame(X_pca, columns=[f'PC{i+1}' for i in range(X_pca.shape[1])])
X_pca_df.head()
```

```python
X_pca_df.to_csv(r'../data/processed/mall_customers_pca.csv', index=False)
```
